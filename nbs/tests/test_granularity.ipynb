# Tests: `fasterai.core.granularity` (source `nbs/core/granularity.ipynb`)

```python
from fastcore.test import *
import torch.nn as nn
from fasterai.core.granularity import *
```

```python
from fastcore.test import *

# get_dim returns correct dimensions for Conv2d
conv = nn.Conv2d(3, 16, 3)
test_eq(Granularities.get_dim(conv, 'weight'), [0])
test_eq(Granularities.get_dim(conv, 'filter'), [2, 3, 4])
test_eq(Granularities.get_dim(conv, 'kernel'), [3, 4])
test_eq(Granularities.get_dim(conv, 'channel'), [2])

# get_dim for Linear
lin = nn.Linear(32, 16)
test_eq(Granularities.get_dim(lin, 'weight'), [0])
test_eq(Granularities.get_dim(lin, 'column'), [1])

# available_modules returns expected types
modules = Granularities.available_modules()
assert nn.Conv2d in modules
assert nn.Linear in modules

# allowed_granularities returns dict
gran = Granularities.allowed_granularities(conv)
assert isinstance(gran, dict)
assert 'weight' in gran
assert 'filter' in gran

# Error: unsupported module type
with ExceptionExpected(ValueError): Granularities.get_dim(nn.ReLU(), 'weight')

# Error: invalid granularity name
with ExceptionExpected(KeyError): Granularities.get_dim(conv, 'nonexistent')
```
